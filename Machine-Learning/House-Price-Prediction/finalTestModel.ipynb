```python
import joblib

rf_model = joblib.load("random_forest_model.pkl")
preprocessor = joblib.load("preprocessor.pkl")
scaler = joblib.load("scaler.pkl")
```

```python
X_test_scaled = joblib.load("X_test_scaled.pkl")
y_test = joblib.load("y_test.pkl")
```

```python
# First property from test data
sample = X_test_scaled[0].reshape(1, -1)

# Actual price
actual_price = y_test.iloc[0]

# Predicted price
predicted_price = rf_model.predict(sample)[0]

# Calculate error
absolute_error = abs(actual_price - predicted_price)
percentage_error = (absolute_error / actual_price) * 100
prediction_accuracy = 100 - percentage_error

print("Actual Price:", actual_price)
print("Predicted Price:", predicted_price)
print("Absolute Error:", absolute_error)
print("Percentage Error:", percentage_error, "%")
print("Prediction Accuracy:", prediction_accuracy, "%")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict all test data
y_pred = rf_model.predict(X_test_scaled)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Average Percentage Error
percentage_error = np.mean(
    np.abs((y_test - y_pred) / y_test)
) * 100

# Approximate accuracy
accuracy = 100 - percentage_error

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
print("Average Percentage Error:", percentage_error, "%")
print("Approximate Prediction Accuracy:", accuracy, "%")
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

comparison["Absolute Error"] = abs(
    comparison["Actual Price"] - comparison["Predicted Price"]
)

comparison.head(10)
```

```python
print("Minimum Actual Price:", y_test.min())
print("Maximum Actual Price:", y_test.max())
print("Mean Actual Price:", y_test.mean())

print("\nR² Score:", r2)

# Check how many predictions are close to actual price
error_percentage = (
    abs(y_test.values - y_pred) / y_test.values
) * 100

print("\nMean Percentage Error:", error_percentage.mean(), "%")
print("Median Percentage Error:", np.median(error_percentage), "%")

print("\nPredictions within 20% error:",
      np.mean(error_percentage <= 20) * 100, "%")
```
